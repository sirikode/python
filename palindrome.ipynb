```python

#without_slicing

s = input("Enter a string: ")

reverse = ""

for char in s:
    reverse = char + reverse

if s == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")

#with_slicing

s = input("Enter a string: ")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")

```
