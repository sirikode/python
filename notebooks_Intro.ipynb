```python
#Count the frequency of each character

text = input("Enter a string: ")

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```

```python
#remove duplicates
text = input("Enter a string: ")

result = ""

for char in text:
    if char not in result:
        result += char

print("String without duplicates:", result)
```
